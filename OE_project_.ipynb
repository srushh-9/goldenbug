```python
# ==============================================
# 0) INSTALL + IMPORTS
# ==============================================
!pip install --quiet GEOparse scanpy scikit-learn pandas numpy scipy matplotlib

import GEOparse
import pandas as pd
import numpy as np
import scanpy as sc
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.linear_model import LogisticRegression, LogisticRegressionCV
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.metrics import roc_auc_score, accuracy_score, f1_score, precision_score, recall_score, roc_curve, auc
from sklearn.pipeline import Pipeline
from collections import defaultdict
import warnings
warnings.filterwarnings("ignore")

DATASETS = ["GSE10950", "GSE25070", "GSE41328", "GSE74602"]

# ==============================================
# 1) HELPER FUNCTIONS
# ==============================================

def detect_probe_and_symbol_columns(gpl_table: pd.DataFrame):
    cols = list(gpl_table.columns)
    probe_candidates = ["ID", "ID_REF", "Probe Set ID", "PROBE_ID", "ProbeID"]
    probe_col = next((c for c in probe_candidates if c in cols), None)
    symbol_candidates = [
        "Gene Symbol", "Symbol", "GENE_SYMBOL", "Gene symbol",
        "gene_assignment", "GeneSymbol"
    ]
    symbol_col = next((c for c in symbol_candidates if c in cols), None)
    if probe_col is None or symbol_col is None:
        raise KeyError(f"Could not detect probe/symbol columns. Found: {cols[:25]}")
    return probe_col, symbol_col


def build_probe_to_symbol_map(gse):
    gpl_name, gpl = list(gse.gpls.items())[0]
    gpl_table = gpl.table.copy()
    probe_col, symbol_col = detect_probe_and_symbol_columns(gpl_table)

    def clean_symbol(x):
        if pd.isna(x):
            return np.nan
        s = str(x)
        for sep in ["///", "//", ";", ",", " "]:
            if sep in s:
                s = s.split(sep)[0]
                break
        s = s.strip()
        if s in ("", "NA", "na", "---", "null", "None"):
            return np.nan
        return s

    mapping = gpl_table[[probe_col, symbol_col]].copy()
    mapping[symbol_col] = mapping[symbol_col].map(clean_symbol)
    mapping = mapping.dropna(subset=[symbol_col])
    return dict(zip(mapping[probe_col].astype(str), mapping[symbol_col].astype(str)))


def extract_sample_expression(gsm):
    tbl = gsm.table
    probe_col = next((c for c in ["ID_REF", "ID", "ProbeID", "PROBE_ID"] if c in tbl.columns), tbl.columns[0])
    value_col = "VALUE" if "VALUE" in tbl.columns else \
                next((c for c in tbl.columns if "signal" in c.lower() or "value" in c.lower()), tbl.columns[-1])
    s = tbl.set_index(probe_col)[value_col]
    s.index = s.index.astype(str)
    s.name = gsm.metadata.get("geo_accession", [""])[0]
    return s


def collapse_to_gene(expr_df, probe2symbol):
    keep_cols = [p for p in expr_df.columns if p in probe2symbol]
    expr = expr_df[keep_cols].copy()
    expr = expr.rename(columns=probe2symbol)
    expr = expr.groupby(level=0, axis=1).mean()
    return expr


def infer_label_from_metadata(gsm):
    texts = []
    for key in ("title", "source_name_ch1", "characteristics_ch1", "description"):
        val = gsm.metadata.get(key, [""])
        if isinstance(val, list):
            texts.extend(val)
        else:
            texts.append(str(val))
    all_text = " | ".join([str(t) for t in texts]).lower()
    tumor_keywords = ["tumor", "cancer", "carcinoma", "adenocarcinoma", "crc"]
    normal_keywords = ["normal", "healthy", "control", "non-tumor", "adjacent normal"]
    if any(k in all_text for k in normal_keywords): return "normal"
    if any(k in all_text for k in tumor_keywords): return "tumor"
    return "unknown"


def load_one_dataset(gse_id):
    print(f"Loading {gse_id} ...")
    gse = GEOparse.get_GEO(geo=gse_id, destdir="./geo_data", annotate_gpl=True)
    probe2symbol = build_probe_to_symbol_map(gse)
    series_list = []
    labels = {}
    for gsm_name, gsm in gse.gsms.items():
        s = extract_sample_expression(gsm)
        series_list.append(s)
        labels[gsm_name] = infer_label_from_metadata(gsm)
    expr = pd.concat(series_list, axis=1).T
    expr_gene = collapse_to_gene(expr, probe2symbol)
    labels = pd.Series(labels).reindex(expr_gene.index)
    print(f"{gse_id}: samples={expr_gene.shape[0]}, genes={expr_gene.shape[1]}")
    return expr_gene, labels


# ==============================================
# 2) LOAD ALL DATASETS
# ==============================================
all_expr, all_labels, batch_ids = [], [], []

for i, gse_id in enumerate(DATASETS):
    expr_gene, labels = load_one_dataset(gse_id)
    all_expr.append(expr_gene)
    all_labels.append(labels)
    batch_ids.extend([i] * expr_gene.shape[0])

# Merge by gene intersection
common_genes = sorted(set.intersection(*map(set, [df.columns for df in all_expr])))
all_expr_trimmed = [df[common_genes] for df in all_expr]
combined_expr = pd.concat(all_expr_trimmed, axis=0)
combined_labels = pd.concat(all_labels, axis=0)
print(f"Merged dataset shape: {combined_expr.shape}")

# ==============================================
# 3) BATCH CORRECTION (ComBat)
# ==============================================
scaler = StandardScaler()
scaled_expr = pd.DataFrame(
    scaler.fit_transform(combined_expr.values),
    index=combined_expr.index, columns=combined_expr.columns
)

adata = sc.AnnData(scaled_expr.values)
adata.obs["batch"] = pd.Series(batch_ids, index=scaled_expr.index).astype(str).values
sc.pp.combat(adata, key="batch")

corrected_expr = pd.DataFrame(adata.X, index=scaled_expr.index, columns=scaled_expr.columns)
labels = combined_labels.fillna("unknown").astype(str)
mask = labels.isin(["tumor", "normal"])

X = corrected_expr.loc[mask]
y = labels.loc[mask].replace({"tumor": 1, "normal": 0}).astype(int)

print("Final dataset:", X.shape, "Labels:", y.value_counts().to_dict())

# ==============================================
# 4) MODEL EVALUATION FRAMEWORK (NO LEAKAGE)
# ==============================================
def evaluate_model(model, X, y):
    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    aucs = cross_val_score(model, X, y, cv=cv, scoring='roc_auc')
    accs = cross_val_score(model, X, y, cv=cv, scoring='accuracy')
    print(f"{model.__class__.__name__}: AUROC={aucs.mean():.3f}±{aucs.std():.3f}, "
          f"Acc={accs.mean():.3f}±{accs.std():.3f}")
    return aucs.mean()

# Reproduced model (like paper): Logistic Regression + L1 (LASSO)
paper_model = Pipeline([
    ("scaler", StandardScaler()),
    ("lasso", LogisticRegressionCV(Cs=10, cv=5, penalty='l1', solver='saga', max_iter=5000))
])

# Custom models
svm_model = Pipeline([
    ("scaler", StandardScaler()),
    ("svm", SVC(kernel='linear', probability=True))
])

rf_model = RandomForestClassifier(n_estimators=200, random_state=42)
gb_model = GradientBoostingClassifier(random_state=42)

# Evaluate
print("\n=== MODEL PERFORMANCE (5-Fold CV) ===")
evaluate_model(paper_model, X, y)
evaluate_model(svm_model, X, y)
evaluate_model(rf_model, X, y)
evaluate_model(gb_model, X, y)

# ==============================================
# 5) ROC COMPARISON VISUALIZATION
# ==============================================
from sklearn.metrics import RocCurveDisplay

X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, test_size=0.2, random_state=42)
pipe = Pipeline([("scaler", StandardScaler()), ("lasso", LogisticRegression(penalty='l1', solver='saga', max_iter=5000))])
pipe.fit(X_train, y_train)
y_pred_prob = pipe.predict_proba(X_test)[:,1]

fpr, tpr, _ = roc_curve(y_test, y_pred_prob)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(6,5))
plt.plot(fpr, tpr, color='blue', lw=2, label=f'Paper LASSO Model (AUC={roc_auc:.3f})')
plt.plot([0,1],[0,1],'--',color='gray')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve – CRC Diagnostic Classifier')
plt.legend(loc="lower right")
plt.show()
```
